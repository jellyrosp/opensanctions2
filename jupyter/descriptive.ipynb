# Descriptive Analysis

```python
from opensanctions2.stats.descriptive.core import (
    count_sanctioned_individual,
    gender_statement_distribution,
    gender_values_distribution,
    gender_datasets_distribution,
    gender_valid_datasets_distribution,
    gender_topics_distribution
    
)

from opensanctions2.stats.descriptive.aux import (
    render_table
)
```

### 1. Total amount of sanctioned sndividuals

```python
total = count_sanctioned_individual()

print(f"Total sanctioned individuals: {total:,}")
```

### 2. Distribution of gender statements 

```python
gender_distribution = gender_statement_distribution()

render_table(gender_distribution)
```

### 3. Distribution of gender values

```python
gender_value_distr = gender_values_distribution()

render_table(gender_value_distr)
```

### 4. Disribution of gender information over datasets

```python
gender_information_datasets = gender_datasets_distribution()

render_table(gender_information_datasets)
```

### 5. Distribution of gender values over valid datasets

```python
gender_information_valid_datasets = gender_valid_datasets_distribution()

render_table(gender_information_valid_datasets)
```

### 6. Distribution of gender values over topics

```python
gender_information_topics = gender_topics_distribution()

render_table(gender_information_topics)
```
