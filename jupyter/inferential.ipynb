# Inferential analysis

## Correlation testing of gender and topics

```python
from opensanctions2.stats.descriptive.core import gender_topics_distribution
from opensanctions2.stats.inferential.core import gender_topic_inference
from opensanctions2.stats.descriptive.aux import remove_relative_frequencies
```

```python
# Load contingency table from your function
contingency_table = remove_relative_frequencies(gender_topics_distribution())
print(contingency_table.to_string(index=False))

# Run the inferential analysis
results_df, chi2_result, standardized_residuals = gender_topic_inference(contingency_table)

# Display summary
print("Global chi-square:", chi2_result)

print("\nResult table:")
display(results_df.style.hide(axis="index"))

print("\nTop 5 topics by log-odds:")
results_df.head().style.hide(axis="index")
```
